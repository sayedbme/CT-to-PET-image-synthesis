```python
import os
from PIL import Image

def process_images(input_folder):
    """
    Recursively process images in a folder, resizing them to 256x256 with 24-bit depth, and overwrite the original images.
    Args:
        input_folder (str): Path to the folder containing the images.
    """
    for root, _, files in os.walk(input_folder):
        for file in files:
            file_path = os.path.join(root, file)
            try:
                with Image.open(file_path) as img:
                    # Check if the file is an image
                    if img.mode not in ('RGB', 'RGBA', 'L'):
                        continue

                    # Convert to RGB (24-bit depth) if not already
                    if img.mode != 'RGB':
                        img = img.convert('RGB')

                    # Resize to 256x256
                    img = img.resize((256, 256), Image.LANCZOS)

                    # Overwrite the original file
                    img.save(file_path, format='JPEG')

                    print(f"Processed and overwritten: {file_path}")
            except Exception as e:
                print(f"Error processing {file_path}: {e}")

# Define your input folder path
input_folder = r"...\Data"

process_images(input_folder)
```
