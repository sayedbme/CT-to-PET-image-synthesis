```python
# ============================================================
# 1. Install Ultralytics
# ============================================================
!pip install -q ultralytics

# ============================================================
# 2. Import Required Libraries
# ============================================================

import os
import torch
import ultralytics

from ultralytics import YOLO
from IPython.display import display, Image


# ============================================================
# 3. Check Environment
# ============================================================

ultralytics.checks()

print("Ultralytics version:", ultralytics.__version__)
print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("Training will run on CPU.")


```

```python
# ============================================================
# 4. Define Dataset and Output Paths
# ============================================================

DATA_YAML = ".../data.yaml"
PROJECT_DIR = ".../results"
RUN_NAME = "epochs"

# ============================================================
# 5. Load YOLO11n Segmentation Model
# ============================================================

# YOLO11n-seg is the YOLO11 Nano model for segmentation.
model = YOLO("yolo11n-seg.pt")

# ============================================================
# 6. Train YOLO11n Segmentation Model
# ============================================================

results = model.train(
    data=DATA_YAML,
    project=PROJECT_DIR,
    name=RUN_NAME,
    imgsz=640,
    batch=8,
    epochs=50,
    optimizer="SGD",
    lr0=0.01,
    momentum=0.9,
    device=0
)

```

```python
# ============================================================
# 7. Display Training Results
# ============================================================

results_path = os.path.join(PROJECT_DIR, RUN_NAME, "results.png")
print("Results saved at:", results_path)
display(Image(filename=results_path, height=400))
```
