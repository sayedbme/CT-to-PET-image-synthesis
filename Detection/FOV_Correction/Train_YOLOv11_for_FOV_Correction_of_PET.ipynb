```python
# ============================================================
# 1. Install Required Packages
# ============================================================
!pip install -q ultralytics roboflow

# ============================================================
# 2. Import Libraries
# ============================================================

import os
import torch
import ultralytics

from roboflow import Roboflow
from ultralytics import YOLO
from IPython.display import display, Image


# ============================================================
# 3. Check Environment
# ============================================================

ultralytics.checks()

print("Ultralytics version:", ultralytics.__version__)
print("PyTorch version:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
else:
    print("Training will run on CPU.")


# ============================================================
# 4. Connect to Roboflow
# ============================================================

rf = Roboflow(api_key="....avC")
rf_project = rf.workspace("....c3tyd").project("roi_from_pet")
version = rf_project.version(1)


# ============================================================
# 5. Download Dataset
# ============================================================

dataset = version.download("yolov11")

# Automatically obtain the path to data.yaml
data_yaml = os.path.join(dataset.location, "data.yaml")

print("Dataset location:", dataset.location)
print("Data YAML:", data_yaml)

```

```python
# ============================================================
# 6. Load YOLO11 Model
# ============================================================

model = YOLO("yolo11n.pt")

# ============================================================
# 7. Define Training Output
# ============================================================

output_project = ".../Detections/ROI_from_PET/results"
run_name = "PET_epochs"


# ============================================================
# 8. Train YOLO11
# ============================================================

results = model.train(
    data=data_yaml,
    project=output_project,
    name=run_name,
    imgsz=640,
    batch=8,
    epochs=50,
    optimizer="SGD",
    lr0=0.01,
    momentum=0.9,
    device=0
)

```

```python
# ============================================================
# 9. Display Training Results
# ============================================================

results_path = os.path.join(output_project, run_name, "results.png")
print("Results saved at:", results_path)
display(Image(filename=results_path,height=400))
```
